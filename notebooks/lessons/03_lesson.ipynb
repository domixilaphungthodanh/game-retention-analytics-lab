## SQL mẫu và giải thích
Mỗi dòng sau JOIN là một lượt thử, nên COUNT(*) không còn là số phiên.

```python
result=read_sql('SELECT p.name,s.session_id,l.name AS level,a.result FROM level_attempts a JOIN sessions s USING(session_id) JOIN players p USING(player_id) JOIN levels l USING(level_id) LIMIT 10;')
result
```

## Thay đổi để quan sát
Dự đoán số dòng, chạy truy vấn thứ hai rồi giải thích điều kiện JOIN/lọc/nhóm.

```python
read_sql('SELECT COUNT(*) AS joined_rows,COUNT(DISTINCT s.session_id) AS sessions FROM sessions s JOIN level_attempts a USING(session_id);')
```
