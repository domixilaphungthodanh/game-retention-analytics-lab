## SQL mẫu và giải thích
Một người mở nhiều phiên vẫn là một người/ngày. Cần COUNT DISTINCT trước khi tính retention.

```python
result=read_sql('SELECT date(started_at) AS activity_date,COUNT(*) AS session_count,COUNT(DISTINCT player_id) AS active_players FROM sessions GROUP BY date(started_at) ORDER BY activity_date;')
result
```

## Thay đổi để quan sát
Dự đoán số dòng, chạy truy vấn thứ hai rồi giải thích điều kiện JOIN/lọc/nhóm.

```python
read_sql('SELECT player_id,date(started_at) AS day,COUNT(*) AS sessions FROM sessions GROUP BY player_id,day HAVING COUNT(*)>1 LIMIT 10;')
```
