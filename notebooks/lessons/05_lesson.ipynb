## SQL mẫu và giải thích
Phải tìm MIN trên lịch sử trước khi lọc nhóm, nếu không người cũ có thể bị nhầm thành người mới.

```python
result=read_sql('WITH firsts AS (SELECT player_id,MIN(date(started_at)) AS cohort_date FROM sessions GROUP BY player_id) SELECT cohort_date,COUNT(*) AS players FROM firsts GROUP BY cohort_date ORDER BY cohort_date;')
result
```

## Thay đổi để quan sát
Dự đoán số dòng, chạy truy vấn thứ hai rồi giải thích điều kiện JOIN/lọc/nhóm.

```python
read_sql("WITH firsts AS (SELECT player_id,MIN(date(started_at)) AS first_day FROM sessions GROUP BY player_id) SELECT player_id,first_day,date(first_day,'+7 days') AS target_d7 FROM firsts LIMIT 10;")
```
