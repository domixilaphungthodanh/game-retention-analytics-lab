## Thay đổi để quan sát
Dự đoán số dòng, chạy truy vấn thứ hai rồi giải thích điều kiện JOIN/lọc/nhóm.

```python
read_sql('SELECT p.name,COUNT(s.session_id) AS sessions FROM players p LEFT JOIN sessions s USING(player_id) GROUP BY p.player_id ORDER BY sessions DESC LIMIT 10;')
```
